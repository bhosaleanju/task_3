## 3. Visual Story 1: Revenue & Profitability Trajectory

## 4. Visual Story 2: The "Discount Trap"
**Objective**: Uncover the relationship between promotional discount rates and net margin health.

```python
fig2 = plot_discount_profit_impact(df, save=False)
plt.show()
```

> **Analytical Insight**:
>- **The 20% Inflection Point**: Transactions with discounts $\le 20\%$ generate strong positive margins (averaging $+20\%$ to $+35\%$).
>- Once discounts exceed $20\%$, profitability plummets into the **Value Destruction Zone** due to unrecovered fixed shipping and handling costs.
>- **Recommendation**: Establish a hard sales discount ceiling at $20\%$, requiring executive approval for any exceptions.

## 5. Visual Story 3: Sub-Category Profitability Matrix
**Objective**: Isolate top-performing profit centers from chronic loss leaders.

```python
fig3 = plot_category_profitability_diverging(df, save=False)
plt.show()
```

> **Analytical Insight**:

>- **Chronic Loss Leaders**: Tables and Bookcases consistently drag down operating performance due to bulky shipping costs and aggressive clearance discounts.

## 6. Visual Story 4: Regional & Category Performance Matrix

## 7. Visual Story 5: Customer Concentration (Pareto 80/20 Rule)
